# Separate flow and reaction timescales

Compare **multirate PHREEQC** with **compositional PHREEQC** over a complete 250-day injection–storage–production cycle, using saved full-duration simulation results.

In the multirate workflow, MRST retains microbial kinetics. Each coarse flow step is followed by local reaction substeps, with PHREEQC aqueous equilibrium after each one. In the compositional workflow, PHREEQC integrates microbial kinetics once after each flow timestep. The equation of state owns gas–liquid partitioning in both workflows.

## Case and schedule

The 50 m column has 20 cells. Injection lasts 50 days, storage lasts 150 days, and production lasts 50 days. Flow timesteps are 2 days; multirate reaction substeps are at most 0.4 days. Both saved cases use the high-rate MET/ACE/SRB parameter set, Søreide–Whitson thermodynamics, carbonate buffering, and 2 wt% dolomite with quartz. Molecular diffusion, dispersion, bacterial diffusion, chemotaxis, and bioclogging are disabled.

These figures use all 125 saved output times, rather than the former two-step setup check. No simulation needs to be rerun to inspect or redraw them.

## Hydrogen consumption throughout the cycle



Dotted lines mark the end of injection (day 50) and the start of production (day 200). Consumption continues during storage. At day 250, the multirate case consumes **46.41%**, compared with **40.18%** for compositional PHREEQC: a difference of **7.55 percentage points**. Methanogenesis (MET) dominates; acetogenesis (ACE) and sulfate reduction (SRB) provide smaller contributions.

Every percentage uses the **total prescribed H₂ injection over the complete cycle** as a fixed denominator: 13,284.903 mol, using nominal ideal-gas conversion at 101325 Pa and 288.15 K. The numerator integrates microbial reaction sources. It excludes dissolution and unrecovered hydrogen and is not an independent well-inventory balance.

## Chemistry histories: pH and inorganic carbon



Lines show the **spatial median across the 20 cells**; shading shows the cellwise minimum–maximum range at each saved time. These are spatial summaries, not well samples or water-mass-weighted concentrations. The pH median is a descriptive statistic, not a mixed-water pH.

Dissolved inorganic carbon (DIC) is total inorganic-carbon molality from PHREEQC, rather than dissolved CO₂ alone. The multirate case develops a lower median pH during storage and a stronger median-DIC response during production. Spatial maps below reveal the fronts that a single summary curve can hide.

## Interpretation and diagnostics

This is a comparison of two configured workflows, not a timestep-convergence study or proof that either kinetic treatment is more accurate. Their kinetic ownership, biomass representation, and flow–reaction splitting differ. The comparison therefore cannot isolate substepping alone.

The compositional run records **98 cell-step elemental-audit failures**, with maximum absolute residual **3.61 × 10⁻⁵ mol**. The multirate equilibrium backend has its elemental audit disabled; its saved EOS inventory checks report no failures. A passing EOS inventory check does not establish elemental conservation through the chemistry handoff.

Both completed simulations retain their full states and packed setups locally. The website publishes only the numerical diagnostics needed to redraw these figures.
