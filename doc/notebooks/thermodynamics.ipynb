## Pressure and salinity comparisons





The pressure panel is a 50 °C slice with zero added salt. PR uses its default binary interactions and is not calibrated to hydrogen–water. ePC-SAFT is supplied as reference data, not a live solver. Henry–Setschenow provides dissolved hydrogen only. Its hydrogen partial pressure is approximated by total pressure in this hydrogen-dominated gas.

The salinity panel fixes 40 °C and 15 MPa. The original RK correlations and flash functions are used unchanged. `solubility_results.mat` stores all computed values. For black-oil PVT generation, see `examplePVTGenerationH2Brine`.

### Temperature-dependent phase behavior
At 15 MPa and zero added salt, compare dissolved hydrogen and water in the gas phase over 1–99 °C. The ePC-SAFT markers are bundled reference data; Henry–Setschenow supplies only liquid hydrogen solubility.
